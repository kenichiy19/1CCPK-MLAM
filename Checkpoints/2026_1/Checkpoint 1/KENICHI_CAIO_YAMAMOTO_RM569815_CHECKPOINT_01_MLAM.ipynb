## IDENTIFICAÇÃO DO ALUNO

### EXECUTE A PRÓXIMA CÉLULA
**NÃO EDITE ESTA CÉLULA. APENAS A EXECUTE.**

```python
import hashlib

nome = input("Digite seu nome: ").strip()
rm = input("Digite seu RM: ").strip()

if nome == "" or rm == "":
    print("ATENÇÃO: preencha nome e RM antes de continuar.")
else:
    base_seed = nome + "|" + rm
    seed = int(hashlib.sha256(base_seed.encode("utf-8")).hexdigest()[:8], 16)
    print("Nome:", nome)
    print("RM:", rm)
    print("Seed individual:", seed)
```

## QUESTÃO 1

Uma equipe registrou a quantidade de atendimentos concluídos em 6 dias consecutivos.

Utilizando os valores armazenados em **dados_q1**, calcule a **média aritmética simples** desse conjunto.

Armazene o resultado na variável:

**resposta_q1**

```python
dados_q1 = [74, 65, 72, 89, 90, 71]
soma_q1 = 74 + 65 + 72 + 89 + 90 + 71
resposta_q1 = soma_q1 / 6
resposta_q1
```

## QUESTÃO 2

Um aluno realizou 4 atividades ao longo do bimestre. Cada atividade possui um peso diferente.

Utilizando os valores armazenados em **valores_q2** e os respectivos pesos em **pesos_q2**, calcule a **média aritmética ponderada**.

Armazene o resultado na variável:

**resposta_q2**

```python
valores_q2 = [6, 8, 9, 9]
pesos_q2 = [2, 4, 4, 5]
soma_q2 = (6 * 2) + (8 * 4) + (9 * 4) + (9 * 5)
sumpeso_q2 = 2 + 4 + 4 + 5
resposta_q2 = soma_q2 / sumpeso_q2
resposta_q2
```

## QUESTÃO 3

Uma empresa registrou a quantidade de pedidos processados em 6 turnos de trabalho.

Utilizando os valores armazenados em **dados_q3**, calcule a **variância populacional** desse conjunto.

Armazene o resultado na variável:

**resposta_q3**

```python
dados_q3 = [51, 82, 94, 47, 68, 88]
media_q3 = (51 + 82 + 94 + 47 + 68 + 88) / 6
# desvio
d1 = 51 - media_q3
d2 = 82 - media_q3
d3 = 94 - media_q3
d4 = 47 - media_q3
d5 = 68 - media_q3
d6 = 88 - media_q3

resposta_q3 = (d1**2 + d2**2 + d3**2 + d4**2 + d5**2 + d6**2) / 6
resposta_q3
```

## QUESTÃO 4

Uma disciplina avaliou 5 entregas de projeto, atribuindo pesos diferentes para cada etapa.

## QUESTÃO 6

Uma unidade registrou o consumo de recursos em 6 medições sucessivas.

Utilizando os valores armazenados em **dados_q6**, calcule:

1. a **variância populacional**;
2. o **desvio padrão populacional**.

Armazene os resultados nas variáveis:

- **resposta_q6_variancia**
- **resposta_q6_desvio**

```python
dados_q6 = [124, 93, 83, 73, 104, 124]
media_q6 = (124 + 93 + 83 + 73 + 104 + 124) / 6

# DESVIO
d1 = 124 - media_q6
d2 = 93 - media_q6
d3 = 83 - media_q6
d4 = 73 - media_q6
d5 = 104 - media_q6
d6 = 124 - media_q6

# VARIANCIA
resposta_q6_variancia = (d1**2 + d2**2 + d3**2 + d4**2 + d5**2 + d6**2) / 6
resposta_q6_desvio = resposta_q6_variancia**0.5
print(resposta_q6_variancia)
print(resposta_q6_desvio)
```
