## 3. Tabelas de distribuição de frequências

Construa ao menos uma tabela para cada tipo de variável identificado. Inclua frequência absoluta, relativa e acumulada. Para variáveis contínuas, utilize distribuição por classes e justifique o número de classes escolhido.

```python
# Variável Qualitativa
tabela_qualitativa = pd.DataFrame()
tabela_qualitativa['Absoluta'] = df['Gender'].value_counts()
tabela_qualitativa['Relativa (%)'] = df['Gender'].value_counts(normalize=True) * 100
tabela_qualitativa['Acumulada'] = tabela_qualitativa['Absoluta'].cumsum()

print(tabela_qualitativa)

# Variável Quantitativa
df['Classes'] = pd.cut(df['Sleep Duration'], bins=9)
tabela_quantitativa = pd.DataFrame()
tabela_quantitativa['Absoluta'] = df['Classes'].value_counts().sort_index()
tabela_quantitativa['Relativa (%)'] = df['Classes'].value_counts(normalize=True).sort_index() * 100
tabela_quantitativa['Acumulada'] = tabela_quantitativa['Absoluta'].cumsum()

print(tabela_quantitativa)
```
